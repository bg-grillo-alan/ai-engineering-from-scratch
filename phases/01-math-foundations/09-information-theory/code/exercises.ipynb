# Exercice 4 

```python
import numpy as np
def perplexity_custom(sequence) :
    ce = [ cross_entropy_loss(index,logits) for (index,logits) in sequence]
    average = np.mean(ce)
    return perplexity(average)

```

```python
sequence = [
    (1, [0.1,2,0.1]),
    (0,[1.5,0.2,0.1]),
    (0,[1.5,0.2,0.1]),
    (0,[1.5,0.2,0.1]),
    (0,[1.5,0.2,0.1]),
    (0,[1.5,0.2,0.1]),
]
perplexity_custom(sequence)
```
