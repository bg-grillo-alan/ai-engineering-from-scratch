```python
# EX 1 part 2 check rotation keeps distance
square = [[0, 0], [1, 0], [1, 1], [0, 1]]
distances = []
transformed = []

## Check distances
for i in range(len(square)) :
    for j in range(i,len(square)) :
        a = square[i]
        b = square[j]
        dist = math.sqrt((b[0]-a[0])**2 + (b[1] - a[1])**2)
        distances.append(dist)
print(distances)

# Rotate
for s in square : 
    rotated = mat_vec_mul(rotation_2d(math.pi/4),s)
    transformed.append(rotated)

## Checl distances again
new_distances = []
for i in range(len(transformed)) :
    for j in range(i,len(transformed)) :
        a = transformed[i]
        b = transformed[j]
        dist = math.sqrt((b[0]-a[0])**2 + (b[1] - a[1])**2)
        new_distances.append(dist)

## Compare distances
one_diff = False
diff_list = []
for d,nd in zip(distances,new_distances) :
    diff_list.append(abs(d-nd) < 1e-9)

print(f"Original points : {square}")
print(f"Distances : {distances}")
print(f"Tranformed : {transformed}")
print(f"New distances : {new_distances}")
print(f" True = same distance with 1e-9 margin: {diff_list}")
```
