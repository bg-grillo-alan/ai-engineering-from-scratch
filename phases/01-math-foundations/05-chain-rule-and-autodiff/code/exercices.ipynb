# Exercice 4


```python
class Dual : 
    def __init__(self,value,derivative=0.0):
        self.value = value
        self.derivative = derivative

    def __mul__(self,other) :
        if isinstance(other,int) : 
            other = Dual(other,0.0)
        nv = self.value * other.value
        nd = self.derivative * other.value + self.value * other.derivative
        return Dual(nv,nd)

    def __rmul__(self, other):
        return self.__mul__(other)

    def __add__(self,other) : 
        nv = self.value + other.value
        nd = self.derivative + other.derivative
        return Dual(nv,nd)

    def __sub__(self,other) :
        nv = self.value - other.value
        nd = self.derivative - other.derivative
        return Dual(nv,nd)
```

```python
x = Dual(3.0,1.0)

y = (x*x*x*x)-3*(x*x)
print(y.value, y.derivative)
```
