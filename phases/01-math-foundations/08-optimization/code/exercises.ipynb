# Exercice 3 

```python
def f(params) : 
    x,y = params
    return x**2 - y**2 

def f_grad(params) :
    x,y = params
    dx = 2*x
    dy = -2*y
    return [dx,dy]

optimizers = [GradientDescent(),SGDMomentum(),Adam()]

for opt in optimizers : 
    history = optimize(opt,f,f_grad,(0.01,0.01))
    print_trajectory("Optimzer",history,f)
```

# Exercice 4 

```python
class GradientDescentDecayed:
    def __init__(self, lr=0.001):
        self.lr = lr

    def step(self, params, grads):
        self.lr = self.lr * 0.999
        return [p - self.lr * g for p, g in zip(params, grads)]

history = optimize(GradientDescentDecayed(),rosenbrock,rosenbrock_gradient,[0,0])
print_trajectory("Decayed lr",history,rosenbrock)
```
