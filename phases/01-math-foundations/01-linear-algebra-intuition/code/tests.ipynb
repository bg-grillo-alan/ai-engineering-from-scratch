```python
scaling = Matrix(rows=[[2,0],[0,3]])
vec = Vector([1,1])
scales = scaling @ vec
scales
```

```python
from random import randint
words_vec = [Vector([randint(-256,256) for _ in range(50) ]) for _ in range(10)]

test = words_vec[0].cosine_similarity(words_vec[0])
print(test)

for i in range(0,9,2):
    print(f"Vec 1 { words_vec[i]}")
    print(f"Vec 2 { words_vec[i+1]}")
    print(words_vec[i].cosine_similarity(words_vec[i+1]))

```
