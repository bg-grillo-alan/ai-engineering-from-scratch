```python
from random import randint
from statistics import median

test_list = [randint(1, 100) for _ in range(9)]
```

```python
test_list.sort()
n = len(test_list)
if n % 2 == 1:
    median_value = test_list[n // 2]
else:
    median_value = (test_list[n // 2 - 1] + test_list[n // 2]) / 2
print(f"Test List: {test_list}")
print(f"Median: {median_value}")
```
